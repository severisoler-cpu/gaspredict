# 01 - Preparacion de Datos

Cargamos el historial de precios (1.2M filas) + macro (Brent, EUR/USD)
y construimos las **features** que el TFT necesita.

```python
import pandas as pd
import numpy as np
import yfinance as yf
from pathlib import Path

DATA_DIR = Path('../data')
print('Cargando precios...')
df = pd.read_csv(DATA_DIR / 'gaspredict_prices.csv.gz', parse_dates=['time'])
print(f'Filas: {len(df):,} | Columnas: {list(df.columns)}')
```

```python
# Solo Gasolina 95 E5
df = df[df['price_gasolina_95_e5'].notna()].copy()
df['date'] = df['time'].dt.date.astype(str)
print(f'Filas con precio 95: {len(df):,}')
df.head(3)
```

```python
# Descargar Brent historico de Yahoo Finance (2024 - hoy)
print('Descargando Brent historico...')
brent = yf.download('BZ=F', start='2024-01-01', auto_adjust=True)['Close'].reset_index()
brent.columns = ['date', 'brent_usd']
brent['date'] = brent['date'].dt.date.astype(str)
eurusd = yf.download('EURUSD=X', start='2024-01-01', auto_adjust=True)['Close'].reset_index()
eurusd.columns = ['date', 'eur_usd']
eurusd['date'] = eurusd['date'].dt.date.astype(str)
macro = pd.merge(brent, eurusd, on='date', how='outer').sort_values('date')
macro['brent_eur'] = macro['brent_usd'] / macro['eur_usd']
macro = macro.ffill()
print(f'Macro: {len(macro)} dias | {macro.date.min()} -> {macro.date.max()}')
macro.head(3)
```

```python
# Feature Engineering
df = df.merge(macro, on='date', how='left')
df['time_dt'] = pd.to_datetime(df['date'])
df['day_of_week'] = df['time_dt'].dt.dayofweek
df['day_of_month'] = df['time_dt'].dt.day
df['month'] = df['time_dt'].dt.month
df['is_weekend'] = (df['day_of_week'] >= 5).astype(int)

df = df.sort_values(['id_station', 'time_dt'])
for lag in [7, 14, 21, 28]:
    df[f'brent_lag_{lag}d'] = df.groupby('id_station')['brent_eur'].shift(lag)
df['price_lag_1d'] = df.groupby('id_station')['price_gasolina_95_e5'].shift(1)
df['price_lag_7d'] = df.groupby('id_station')['price_gasolina_95_e5'].shift(7)

df = df.dropna(subset=['price_gasolina_95_e5', 'brent_usd', 'price_lag_1d'])
print(f'Dataset final: {len(df):,} filas')
df.describe()
```

```python
# Filtrar ruta y guardar en Parquet
route_mask = df['municipality'].str.contains('Riba|Manises|Mislata|Benag', case=False, na=False)
df_route = df[route_mask].copy()
df_full = df.copy()
print(f'Ruta: {df_route.id_station.nunique()} estaciones | {len(df_route):,} filas')
print(f'Full Valencia: {df_full.id_station.nunique()} estaciones | {len(df_full):,} filas')
df_route.to_parquet(DATA_DIR / 'train_route.parquet', index=False)
df_full.to_parquet(DATA_DIR / 'train_full.parquet', index=False)
print('Guardado! Pasa al notebook 02_train_tft')
```
